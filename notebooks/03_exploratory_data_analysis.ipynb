# Exploratory Data Analysis

## Business Objective
Analyse customer repayment behaviour and identify high-risk customer groups.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('../data/processed/creditguard_cleaned.csv')
df.head()
```

## Data Validation

```python
df.info()
```

## Target Analysis

```python
df['default_payment_next_month'].value_counts(normalize=True)
```

## Demographic Analysis

```python
df.groupby('age_group')['default_payment_next_month'].mean().sort_values(ascending=False)
```

## Credit-Limit Analysis

```python
df.groupby('credit_limit_group')['default_payment_next_month'].mean().sort_values(ascending=False)
```

## Repayment-History Analysis

```python
df.groupby('maximum_delay_months')['default_payment_next_month'].mean().sort_values(ascending=False)
```

## Bill and Payment Analysis

```python
df.groupby('default_payment_next_month')[['average_bill_amount', 'average_payment_amount']].mean()
```

## High-Risk Segment Analysis

```python
segments = df.groupby(['age_group', 'credit_limit_group', 'education_label', 'has_payment_delay']).agg(total_customers=('id', 'count'), defaulters=('default_payment_next_month', 'sum')).reset_index()
segments['default_rate'] = segments['defaulters'] / segments['total_customers']
segments[segments['total_customers'] >= 100].sort_values(by='default_rate', ascending=False).head(10)
```

## Key Business Insights
1. First Missed Payment (PAY_0) is Critical.
2. Credit Limit is a strong Proxy for Risk.
3. High accumulated delay months strongly correlate with default.
4. Specific combinations of age and credit limit represent the highest risk clusters.

## Conclusions & Next Phase
The EDA confirms that payment delay history and credit limits are highly predictive. The next phase will focus on building machine learning models to automate these predictions.
